# CDF to PDF

The CDF is $F(x)=P(X\leq x)$.


## Q1. Given an array of CDF values, calculate the PDF

`x = [0, 1, 2, 3, 4]`, `cdf = [0, 0.1, 0.5, 0.9, 1.0]`.

### Mathematical derivation: finite differences

For a continuous random variable,

$$F(x)=\int_{-\infty}^x f(t)\,dt,\qquad f(x)=F'(x).$$

Given only sampled CDF values, calculate the average density over each interval:

$$\bar f_i=\frac{F(x_i)-F(x_{i-1})}{x_i-x_{i-1}}.$$

| Interval | CDF difference | Average PDF |
|---|---:|---:|
| 0–1 | 0.1 | 0.1 |
| 1–2 | 0.4 | 0.4 |
| 2–3 | 0.4 | 0.4 |
| 3–4 | 0.1 | 0.1 |


**Note: here calculation methods for PDF and PMF are the same**


```python
# array

def cdf_to_pdf(x, cdf):
    pdf = []

    for i in range(1, len(x)):
        dx = x[i] - x[i-1]
        density = (cdf[i] - cdf[i-1]) / dx
        pdf.append(round(density, 2))

    return pdf

x = [0, 1, 2, 3, 4]
cdf = [0, 0.1, 0.5, 0.9, 1.0]
print(cdf_to_pdf(x, cdf))
```

```python
# df
import pandas as pd

df = pd.DataFrame({'x': x, 'cdf': cdf})

df['pdf'] = (df['cdf'].diff() / df['x'].diff()).round(2)

df
```

## Q2. Given a normal CDF, calculate the PDF at x


### Method 1: Use Python Package: SciPy

Syntax: `norm.pdf(x, loc=0, scale=1)`

* x: point at which to calculate PDF
* loc: mean
* scale: standard deviation (not variance)

```python
from scipy.stats import norm

print(norm.pdf([-1, 0, 1], loc=0, scale=1))
```

### Method 2: Numerical differentiation of the CDF function

If we can evaluate the CDF function but do not know its analytical derivative, use the central difference approximation:

$$f(x)\approx\frac{F(x+h)-F(x-h)}{2h}.$$

```python
def cdf_to_pdf_at_x(cdf_func, x, h=1e-5):
    return (cdf_func(x+h) - cdf_func(x-h)) / (2*h)

print(cdf_to_pdf_at_x(norm.cdf, 0))
```
