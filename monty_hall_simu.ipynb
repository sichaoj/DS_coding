# Monty Hall Problem 🐐 🏎

## Q. Should you stay or switch?

There are three doors. One hides a car and the other two hide goats.

1. The player randomly chooses one door.
2. The host, who knows where the car is, always opens another door containing a goat.
3. The player can stay with the original choice or switch to the remaining unopened door.

Which strategy maximizes the probability of winning the car?

## Method 1: Full simulation

Simulate the car location, the player's initial choice, the host's action, and both strategies. The host chooses randomly among the eligible goat doors if there are two.

```python
import random

random.seed(42)

n = 100000
stay_wins = 0
switch_wins = 0

for i in range(n):

    car = random.randint(1, 3)
    choice = random.randint(1, 3)

    if choice == car:
        stay_wins += 1
    else:
        switch_wins += 1

print("Stay win rate:", stay_wins / n)
print("Switch win rate:", switch_wins / n)
```
